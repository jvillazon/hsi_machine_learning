```python
df_condition = df.copy()
df_condition['Sample ID'] = df_condition['Sample ID'].astype(int)
```

```python
choices = ['Ctrl/Veh', 'Ctrl/SGLT2i', 'WD/Veh', 'WD/SGLT2i','Ctrl/Veh', 'Ctrl/SGLT2i', 'WD/Veh', 'WD/SGLT2i']
df_condition["Condition"] = np.select(conditions, choices, default="Unknown")
```
